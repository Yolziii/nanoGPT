# Test Kobzar BPE tokenizer

Small helper notebook for manual encode/decode checks.

```python
import sys
if hasattr(sys.stdout, "reconfigure"):
    sys.stdout.reconfigure(encoding="utf-8")

from pathlib import Path
from tokenizers import Tokenizer

DATA_DIR = Path.cwd()
TOKENIZER_PATH = DATA_DIR / "kobzar-bpe" / "tokenizer.json"

tokenizer = Tokenizer.from_file(str(TOKENIZER_PATH))

print(f"Loaded: {TOKENIZER_PATH}")
print(f"Vocab size: {tokenizer.get_vocab_size()}")
print(f"<|unknown|>: {tokenizer.token_to_id('<|unknown|>')}")
print(f"<|endoftext|>: {tokenizer.token_to_id('<|endoftext|>')}")
print(f"<|endofpart|>: {tokenizer.token_to_id('<|endofpart|>')}")
```

```python
def encode(text: str) -> list[int]:
    return tokenizer.encode(text).ids


def decode(ids: list[int]) -> str:
    return tokenizer.decode(ids, skip_special_tokens=False)


def show_tokens(text: str, max_tokens: int = 100):
    encoded = tokenizer.encode(" " + text)
    rows = list(zip(encoded.ids, encoded.tokens))
    print(f"Text length: {len(text)}")
    print(f"Token count: {len(encoded.ids)}")
    print(f"Compression rate: {len(text) / len(encoded.ids):.4f}" if encoded.ids else "Compression rate: n/a")
    print()

    shown_rows = rows[:max_tokens]
    for token_id, token in shown_rows:
        visible = repr(token)[1:-1]
        print(f"{token_id:>5}  {visible}")

    if len(rows) > max_tokens:
        print(f"... clipped: shown {max_tokens} of {len(rows)} tokens")

    return encoded
```

```python
sample_text = "Реве та стогне Дніпр широкий,\nсердитий вітер завива."

encoded = show_tokens(sample_text)
ids = encoded.ids
decoded = decode(ids)

print()
print("ids:", ids)
print("decoded:", repr(decoded))
```

```python
# Edit this cell for quick manual experiments.
text = "2015, Oxford Dictionaries named the emoji U+1F602 😂 FACE WITH TEARS OF JOY its word of the year.[11][12]"

encoded = show_tokens(text)
print()
print("decoded:", repr(decode(encoded.ids)))
```
